```python
#202432036_TristaYosevaSitorus
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error

warnings.filterwarnings('ignore')

SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Perangkat komputasi:', device)
print('Versi PyTorch      :', torch.__version__)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>

```python
#202432036_TristaYosevaSitorus

df = pd.read_csv('SeoulBikeData.csv', encoding='cp1252') # pakai encoding cp1252 karena nama kolom menggunakan simbol derajat (°C)

print('Ukuran dataset (baris, kolom):', df.shape)
print('\nNama kolom:')
print(list(df.columns))
print('\nTipe data dan jumlah nilai non-null:')
df.info()
print('\nJumlah nilai hilang per kolom:')
print(df.isna().sum())
print('\nJumlah baris duplikat:', df.duplicated().sum())
```

```python
#202432036_TristaYosevaSitorus

display(df.head())
```

```python
#202432036_TristaYosevaSitorus

display(df.describe().T)
```

```python
#202432036_TristaYosevaSitorus

df['Datetime'] = pd.to_datetime(df['Date'], format='%d/%m/%Y') + pd.to_timedelta(df['Hour'], unit='h')
df = df.sort_values('Datetime').reset_index(drop=True)

selisih_waktu = df['Datetime'].diff().dropna()
print('Rentang waktu            :', df['Datetime'].min(), 's.d.', df['Datetime'].max())
print('Seluruh selisih = 1 jam? :', (selisih_waktu == pd.Timedelta(hours=1)).all())
```

```python
#202432036_TristaYosevaSitorus

print(df['Functioning Day'].value_counts())
print('\nNilai Rented Bike Count saat Functioning Day = No:',
      df.loc[df['Functioning Day'] == 'No', 'Rented Bike Count'].unique())
```

```python
#202432036_TristaYosevaSitorus

df['Holiday_flag'] = (df['Holiday'] == 'Holiday').astype(int)
df['Functioning_flag'] = (df['Functioning Day'] == 'Yes').astype(int)

df['Hour_sin'] = np.sin(2 * np.pi * df['Hour'] / 24)
df['Hour_cos'] = np.cos(2 * np.pi * df['Hour'] / 24)
df['DayOfWeek'] = df['Datetime'].dt.dayofweek
df['DayOfWeek_sin'] = np.sin(2 * np.pi * df['DayOfWeek'] / 7)
df['DayOfWeek_cos'] = np.cos(2 * np.pi * df['DayOfWeek'] / 7)
df['Is_Weekend'] = (df['DayOfWeek'] >= 5).astype(int)

print('Korelasi Temperature vs Dew point temperature:',
      round(df['Temperature(°C)'].corr(df['Dew point temperature(°C)']), 3))

fitur_numerik = ['Rented Bike Count', 'Temperature(°C)', 'Humidity(%)', 'Wind speed (m/s)',
                 'Visibility (10m)', 'Solar Radiation (MJ/m2)', 'Rainfall(mm)', 'Snowfall (cm)']
fitur_lain = ['Hour_sin', 'Hour_cos', 'DayOfWeek_sin', 'DayOfWeek_cos',
              'Is_Weekend', 'Holiday_flag', 'Functioning_flag']
kolom_fitur = fitur_numerik + fitur_lain
IDX_TARGET = kolom_fitur.index('Rented Bike Count')

print('\nJumlah fitur input per jam:', len(kolom_fitur))
print(kolom_fitur)
```

```python
#202432036_TristaYosevaSitorus

WINDOW = 24
n_sampel = len(df) - WINDOW
idx_latih_akhir = int(n_sampel * 0.70)
idx_val_akhir = int(n_sampel * 0.85)

baris_latih = idx_latih_akhir + WINDOW
scaler_fitur = MinMaxScaler().fit(df.loc[:baris_latih - 1, fitur_numerik])
scaler_target = MinMaxScaler().fit(df.loc[:baris_latih - 1, ['Rented Bike Count']])

data_fitur = df[kolom_fitur].copy()
data_fitur[fitur_numerik] = scaler_fitur.transform(df[fitur_numerik])
target_skala = scaler_target.transform(df[['Rented Bike Count']]).ravel()

print('Maks Rented Bike Count pada data latih (dasar scaler):', int(scaler_target.data_max_[0]))
print('Periode target latih     :', df['Datetime'].iloc[WINDOW], 's.d.', df['Datetime'].iloc[idx_latih_akhir + WINDOW - 1])
print('Periode target validasi  :', df['Datetime'].iloc[idx_latih_akhir + WINDOW], 's.d.', df['Datetime'].iloc[idx_val_akhir + WINDOW - 1])
print('Periode target uji       :', df['Datetime'].iloc[idx_val_akhir + WINDOW], 's.d.', df['Datetime'].iloc[-1])
```

```python
#202432036_TristaYosevaSitorus

def buat_sequences(fitur, target, window):
    X, y = [], []
    for i in range(len(fitur) - window):
        X.append(fitur[i:i + window])
        y.append(target[i + window])
    return np.array(X, dtype=np.float32), np.array(y, dtype=np.float32)

X_all, y_all = buat_sequences(data_fitur.values, target_skala, WINDOW)
waktu_target = df['Datetime'].values[WINDOW:]
operasi_target = df['Functioning_flag'].values[WINDOW:] == 1

def ambil_bagian(awal, akhir):
    sl = slice(awal, akhir)
    valid = operasi_target[sl]
    return X_all[sl][valid], y_all[sl][valid], waktu_target[sl][valid]

X_train, y_train, waktu_train = ambil_bagian(0, idx_latih_akhir)
X_val, y_val, waktu_val = ambil_bagian(idx_latih_akhir, idx_val_akhir)
X_test, y_test, waktu_test = ambil_bagian(idx_val_akhir, n_sampel)

print('Total sampel sebelum penyaringan:', len(X_all))
print('X_train:', X_train.shape, '| y_train:', y_train.shape)
print('X_val  :', X_val.shape, '| y_val  :', y_val.shape)
print('X_test :', X_test.shape, '| y_test :', y_test.shape)
print('Bentuk X = (jumlah_sampel, time_step=24, jumlah_fitur) sesuai batch_first=True pada nn.LSTM')
```

```python
#202432036_TristaYosevaSitorus

X_train_t, y_train_t = torch.tensor(X_train), torch.tensor(y_train).unsqueeze(1)
X_val_t, y_val_t = torch.tensor(X_val), torch.tensor(y_val).unsqueeze(1)
X_test_t, y_test_t = torch.tensor(X_test), torch.tensor(y_test).unsqueeze(1)

BATCH_SIZE = 64

train_loader = DataLoader(TensorDataset(X_train_t, y_train_t), batch_size=BATCH_SIZE, shuffle=True)
val_loader = DataLoader(TensorDataset(X_val_t, y_val_t), batch_size=BATCH_SIZE, shuffle=False)

print('Jumlah batch latih:', len(train_loader), '| jumlah batch validasi:', len(val_loader))
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432036_TristaYosevaSitorus

class SequenceModel(nn.Module):
    def __init__(self, input_size, hidden_size=64, num_layers=2, dropout=0.2):
        super(SequenceModel, self).__init__()
        self.lstm = nn.LSTM(input_size=input_size, hidden_size=hidden_size,
                            num_layers=num_layers, batch_first=True, dropout=dropout)
        self.dropout = nn.Dropout(dropout)
        self.fc = nn.Linear(hidden_size, 1)

    def forward(self, x):
        out, (h_n, c_n) = self.lstm(x)
        terakhir = out[:, -1, :]
        return self.fc(self.dropout(terakhir))
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432036_TristaYosevaSitorus

epochs = 100
learning_rate = 0.001
hidden_size = 64
num_layers = 2
dropout = 0.2
patience = 10

model = SequenceModel(input_size=X_train.shape[2], hidden_size=hidden_size,
                      num_layers=num_layers, dropout=dropout).to(device)
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=learning_rate)
print(model)
print('Jumlah parameter yang dilatih:', sum(p.numel() for p in model.parameters() if p.requires_grad))


train_losses = []
val_losses = []


terbaik_val = float('inf')
state_terbaik = None
tunggu = 0

for epoch in range(1, epochs + 1):
    model.train()
    total_latih = 0.0
    for xb, yb in train_loader:
        xb, yb = xb.to(device), yb.to(device)
        optimizer.zero_grad()
        loss = criterion(model(xb), yb)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)  # cegah gradien meledak
        optimizer.step()
        total_latih += loss.item() * xb.size(0)
    train_losses.append(total_latih / len(train_loader.dataset))

    model.eval()
    total_val = 0.0
    with torch.no_grad():
        for xb, yb in val_loader:
            xb, yb = xb.to(device), yb.to(device)
            total_val += criterion(model(xb), yb).item() * xb.size(0)
    val_losses.append(total_val / len(val_loader.dataset))

    if val_losses[-1] < terbaik_val:
        terbaik_val = val_losses[-1]
        state_terbaik = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
        tunggu = 0
    else:
        tunggu += 1

    if epoch % 5 == 0 or epoch == 1:
        print(f'Epoch {epoch:02d}/{epochs} | Train Loss: {train_losses[-1]:.6f} | Val Loss: {val_losses[-1]:.6f}')
    if tunggu >= patience:
        print(f'EarlyStopping aktif pada epoch {epoch}.')
        break

model.load_state_dict(state_terbaik)
print(f'\nVal loss terbaik: {terbaik_val:.6f} pada epoch {int(np.argmin(val_losses)) + 1}')
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">5. VISUALISASI LOSS TRAINING & PERBANDINGAN PREDIKSI VS AKTUAL</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menampilkan grafik konvergensi nilai Loss terhadap Epoch dan grafik evaluasi perbandingan hasil prediksi model terhadap data aktual.</p>
</div>

```python
#202432036_TristaYosevaSitorus

model.eval()
with torch.no_grad():
    pred_skala = model(X_test_t.to(device)).cpu().numpy()

y_pred = scaler_target.inverse_transform(pred_skala).ravel()
y_pred = np.clip(y_pred, 0, None)
y_aktual = scaler_target.inverse_transform(y_test_t.numpy()).ravel()

print('Jumlah sampel uji:', len(y_aktual))
print('Contoh 5 pertama (aktual vs prediksi):')
print(np.round(np.c_[y_aktual[:5], y_pred[:5]], 1))
```

```python
#202432036_TristaYosevaSitorus

plt.figure(figsize=(10, 4))
plt.plot(train_losses, color='#114F6A', linewidth=2, label='Training Loss')
plt.plot(val_losses, color='#1FA4B1', linewidth=2, linestyle='--', label='Validation Loss')
plt.title('Grafik Penurunan Loss Selama Pelatihan')
plt.xlabel('Epoch')
plt.ylabel('Loss (MSE, data terskala)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()

def dengan_celah(waktu, nilai):
    posisi = np.where(np.diff(waktu) > np.timedelta64(1, 'h'))[0] + 1
    return np.insert(waktu, posisi, waktu[posisi - 1]), np.insert(nilai, posisi, np.nan)

w_akt, v_akt = dengan_celah(waktu_test, y_aktual)
w_prd, v_prd = dengan_celah(waktu_test, y_pred)
plt.figure(figsize=(12, 5))
plt.plot(w_akt, v_akt, color='#114F6A', linewidth=1.2, label='Data Aktual')
plt.plot(w_prd, v_prd, color='#1FA4B1', linewidth=1.2, label='Prediksi LSTM')
plt.title('Perbandingan Rented Bike Count Aktual vs Prediksi LSTM pada Data Uji')
plt.xlabel('Waktu')
plt.ylabel('Rented Bike Count (sepeda/jam)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()

n_zoom = 168
w_akt, v_akt = dengan_celah(waktu_test[:n_zoom], y_aktual[:n_zoom])
w_prd, v_prd = dengan_celah(waktu_test[:n_zoom], y_pred[:n_zoom])
plt.figure(figsize=(12, 5))
plt.plot(w_akt, v_akt, color='#114F6A', linewidth=1.8, marker='o', markersize=3, label='Data Aktual')
plt.plot(w_prd, v_prd, color='#1FA4B1', linewidth=1.8, marker='o', markersize=3, label='Prediksi LSTM')
plt.title('Aktual vs Prediksi LSTM (168 Jam Pertama Data Uji)')
plt.xlabel('Waktu')
plt.ylabel('Rented Bike Count (sepeda/jam)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">6. EVALUASI MODEL (MAE, MSE, RMSE)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menghitung metrik regresi hanya pada data uji yang tidak pernah dipakai saat pelatihan, dan membandingkannya dengan baseline sederhana.</p>
</div>

```python
#202432036_TristaYosevaSitorus

mae = mean_absolute_error(y_aktual, y_pred)
mse = mean_squared_error(y_aktual, y_pred)
rmse = np.sqrt(mse)

hasil_lstm = pd.DataFrame({
    'Metrik': ['MAE', 'MSE', 'RMSE'],
    'Nilai': [mae, mse, rmse],
    'Satuan': ['sepeda/jam', '(sepeda/jam)^2', 'sepeda/jam']
})
print('HASIL EVALUASI MODEL LSTM PADA DATA UJI')
display(hasil_lstm.round(2))


y_base_jam_lalu = scaler_target.inverse_transform(X_test[:, -1, [IDX_TARGET]]).ravel()
y_base_kemarin = scaler_target.inverse_transform(X_test[:, 0, [IDX_TARGET]]).ravel()

def ringkas(nama, pred):
    return {'Model': nama,
            'MAE': mean_absolute_error(y_aktual, pred),
            'RMSE': np.sqrt(mean_squared_error(y_aktual, pred))}

perbandingan = pd.DataFrame([
    ringkas('LSTM', y_pred),
    ringkas('Baseline (jam sebelumnya)', y_base_jam_lalu),
    ringkas('Baseline (jam sama kemarin)', y_base_kemarin),
])
print('\nPERBANDINGAN DENGAN BASELINE')
display(perbandingan.round(2))
```
